```python
#generate a list of 100 even numbers

even=[]
number=0
while True:
    if len(even)>100:
        break
    number=number+1
    if number%2==0:
        even.append(number)
even
```

```python
def get_even_numbers(count):
    even=[]
    number=0
    while True:
        if len(even)>count:
            break
        number=number+1
        if number%2==0:
            even.append(number)
    return even

n1=get_even_numbers(100)
n1
```

```python
#you are given a list find the highest numebr in that list
l1=[2,4,6,3,5]
def get_max(numbers:list)->int:
  high=numbers[0]
  for i in numbers:
    if high<i:
      high=i
  return high
get_max(l1)
```
