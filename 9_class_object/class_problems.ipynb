```python
class Circle:
    pi=3.14

    def __init__(self,radius):
        self.radius=radius
    
    def calcualte_area(self):
        return self.pi * (self.radius**2)

    def calculate_circumference(self):
        return 2 * self.pi * self.radius


c1=Circle(5)
c2=Circle(10)

c2.calculate_circumference()
c2.calcualte_area()
```

```python
class Bank:
    bank_name="HDFC"
    def __init__(self,account_numbner,account_name,balance):
        self.account_numbner=account_numbner
        self.account_name=account_name
        self.balance=balance

    def get_info(self):
        return f"Account Number: {self.account_numbner}, Account Name: {self.account_name}, Balance: {self.balance}"

ac1=Bank(123456,"John Doe",1000)
ac1.get_info()

ac2=Bank(789012,"Jane Smith",2000)
ac2.get_info()
```
